### Ejercicio 1: Análisis de una expresión algebraica

Una expresión matemática se recibe como una **cadena** de caracteres. Definir una función `analizar_expresion(expr)` que retorne una **tupla** `(balanceada, n_operadores, variables)` donde:

- `balanceada` es `True` si los paréntesis `( )` y corchetes `[ ]` están correctamente balanceados y anidados, y `False` en caso contrario. (Ayuda: usar una **lista como pila** y un **diccionario** que asocie cada símbolo de cierre con su símbolo de apertura.)
- `n_operadores` es la cantidad de operadores `+ - * / ^` que aparecen en la expresión.
- `variables` es una **lista ordenada** con las letras distintas que aparecen en la expresión (por ejemplo `['x', 'y']`).

Probar con las siguientes expresiones e imprimir el resultado de cada una con un mensaje del tipo:

```
(x+1)*(x-1) -> balanceada: True, operadores: 3, variables: ['x']
```

````python
expresiones = ["(x+1)*(x-1)", "[(2*x)+3]/y", "((2*x)+3", "3*x^2-(y+4))", "[(x+1])"]
````

````python
# Solución:
simbolos = {
    "]" : "[",
    ")" : "(",
}

operaciones = ["+", "-", "*", "/", "^"]

def expr(texto):
    pila = [ ]
    op = 0
    var = []
    b = True

    for letra in texto:
        if letra in simbolos.values():
            pila.append(letra)
        elif letra in simbolos.keys():
            if len(pila) == 0 or pila[-1]!= simbolos[letra]: # Cerrar ([ en el orden que aparezcan
                b =  False
            else:
                pila.pop()
        elif letra in operaciones:
                op += 1
        elif letra not in var and letra.isalpha():
            var.append(letra)
    if len(pila) != 0:
        b = False
    return (b, op, var)

for texto in expresiones:
    resultado = expr(texto)
    print(f"Expresión: {texto} -> Resultado: {resultado}")
````

### Ejercicio 2: Estadística descriptiva de una muestra

Se tiene una muestra de datos en una lista. **Sin usar el módulo `statistics`** para el cálculo:

- Definir `estadisticas(datos)` que retorne una **tupla** `(media, mediana, varianza, desviacion)`, usando la **varianza muestral** (dividiendo entre $n-1$). La mediana debe calcularse **sin modificar** la lista original.
- Con una **lista por comprensión**, crear `z` con los puntajes estandarizados $z_i = (x_i - \bar{x})/s$, redondeados a 4 decimales.
- Crear la lista `atipicos` con los datos cuyo $|z_i| > 2$ (puede usar `zip`).
- Verificar los resultados comparando con `statistics.mean`, `statistics.median` y `statistics.variance` usando `math.isclose`.

Imprimir los resultados y comprobar que `datos` conserva su orden original.

````python
import math
import statistics

datos = [12, 15, 14, 10, 13, 16, 11, 14, 45, 13]
````

````python
# Solución:
def estadisticas(datos):
    # Media
    media = sum(datos)/len(datos)

    # Mediana
    if len(datos)%2==0:
        aux1 = int(len(datos)/2)
        aux2 = int(len(datos)/2) +1
    else:
        aux1 = aux2 = int(len(datos)/2) + 1
    centro = sorted(datos)[aux1 - 1 : aux2]
    mediana = sum(centro)/len(centro)

    # Varianza
    var = 0
    for n in datos:
        var += ( n - media)**2
    var = var / (len(datos)-1)
        
    return [media, mediana, var, math.sqrt(var)]

media, mediana, var, s = estadisticas(datos)
datos_centalizados = [round((x - media)/s,4) for x in datos]
atipicos = [z for z in datos_centalizados if abs(z)>2]

if math.isclose(media,statistics.mean(datos)) and math.isclose(mediana,statistics.median(datos)) and math.isclose(var,statistics.variance(datos)):
    print("Comprobación realizada: Profe lo logré c:")
else:
    print("Rotundo fracaso: Profe le fallé :c")
print("\n--Resultados--")
print("Media: ", media)
print("Mediana: ", mediana)
print("Varianza: ", var)
print("Desviancion: ", s)
print("Valores centralizados:", datos_centalizados)
print("Valores atipicos:", atipicos)

# datos conserva su orden original
print("\nDatos = ", datos)
````

### Ejercicio 3: Números complejos como tuplas

Un número complejo $z = a + bi$ se representa como la **tupla** `(a, b)`. Sin usar el tipo `complex` de Python, definir:

- `producto(z, w)` que retorne el producto de dos complejos como tupla. (Desempaquetar: `a, b = z`.)
- `modulo(z)` que retorne $|z|$.
- `a_cadena(z)` que retorne el número como **cadena** con el formato `"5+5i"` o `"3-2i"` (el signo de la parte imaginaria debe escribirse correctamente).
- `potencia(z, n)` que retorne $z^n$ (con $n \geq 0$ entero) multiplicando repetidamente con `producto`.

Pruebas:

1. Imprimir $(1+2i)(3-i)$ en formato cadena.
2. Calcular $(1+i)^8$ e $i^4$, e imprimirlos. Verificar con `math.isclose` que $|(1+i)^8| = |1+i|^8$.
3. Dada la lista `zs`, imprimir cada número en formato cadena junto con su módulo, **ordenados por módulo creciente**.

````python
import math

zs = [(1, 2), (3, -1), (0, 1), (-2, -2)]
````

````python
# Solución:
def producto(z,w):
    a , b = z
    c, d = w
    return [a*c - b*d, a*d + b*c]

def modulo(z):
    a , b = z
    return math.sqrt(a**2 + b**2)

def a_cadena(z):
    a,b = z
    if b < 0:
        return f"{a} - {abs(b)}i"
    if b > 0:
        return f"{a} + {b}i"
    else:                       # b == 0
        return f"{a}"

def potencia(z, n):
    res = (1, 0)
    for i in range(n):
        res = producto(z, res)
    return res

# Pruebas
print("Prueba 1")
print(a_cadena(producto((1,2), (3, -1))))

print("\nPrueba 2")
print(a_cadena(potencia((1,1), 8)))
print(a_cadena(potencia((0,1), 4)))
print("Verificacion:", math.isclose(modulo(potencia((1,1), 8)), modulo((1,1))**8))

print("\nPrueba 3")
zs.sort(key = modulo)
print("Numero \t\t Modulo")
for num in zs:
    print(f"{a_cadena(num)}\t\t{modulo(num)}")
````

### Ejercicio 4: Polinomios como diccionarios

Un polinomio se puede representar de forma compacta con un **diccionario** `{exponente: coeficiente}` que guarda solo los términos no nulos. Por ejemplo, $p(x) = 3x^5 - 2x^2 + 7$ es `{5: 3, 2: -2, 0: 7}`.

Definir las siguientes funciones:

- `evaluar(p, x)` — evalúa el polinomio en `x`.
- `derivar(p)` — retorna un **nuevo diccionario** con la derivada de `p` (los términos constantes desaparecen).
- `sumar(p, q)` — retorna un nuevo diccionario con $p+q$, **eliminando** los términos cuyo coeficiente resulte 0. No modificar `p` ni `q`.
- `a_cadena(p)` — retorna el polinomio como **cadena**, con los términos en orden decreciente de exponente. Reglas: el exponente 1 se escribe `x`, el exponente 0 solo el coeficiente, y el coeficiente siempre se escribe (aunque sea 1). Ejemplo: `{4: 15, 1: -4}` → `"15x^4 - 4x"`. El polinomio vacío es `"0"`.

Pruebas con `p = {5: 3, 2: -2, 0: 7}` y `q = {5: -3, 1: 4, 0: -7}`:

1. Imprimir `p`, `p'` y `p + q` en formato cadena.
2. Evaluar `p(2)` y `p'(1)`.
3. Verificar numéricamente que $(p+q)(3) = p(3) + q(3)$.

````python
p = {5: 3, 2: -2, 0: 7}
q = {5: -3, 1: 4, 0: -7}
````

````python
# Solución:
def evaluar(p,x):
    potencias = list(p.keys())
    coeficientes = list(p.values())
    return sum([coeficientes[i]*x**potencias[i] for i in range(len(p))])

def derivar(p):
    potencias = [n - 1 for n in list(p.keys()) if n-1>=0 ]
    coeficientes = [list(p.values())[i] * list(p.keys())[i] for i in range(len(potencias))]
    return dict(zip(potencias, coeficientes))

def sumar(p, q):
    suma = {}
    for exp, coef in p.items():
        suma[exp] = coef

    for exp, coef in q.items():
        if exp in suma:
            suma[exp] += coef
        else:
            suma[exp] = coef
    for exp in list(suma):
        if suma[exp] == 0:
            del suma[exp]
    return suma
  
def a_cadena(p):
    texto = ""
    for i in range(len(p)):
        coeficiente = list(p.values())[i]
        if(coeficiente >0) and i!=0:
            texto += f"+{coeficiente}"
        else:
            texto += f"{coeficiente}"
        potencia = list(p.keys())[i]
        if potencia != 0:
            texto += "x"
            if potencia !=1:
                texto += f"^{list(p.keys())[i]}"
    return texto

# Pruebas 
print("Prueba 1")
print("p:\t",a_cadena(p))
print("p':\t",a_cadena(derivar(p)))
print("p + q:\t",a_cadena(sumar(p,q)))
print("\nPrueba 2")
print("p(2):\t", evaluar(p, 2))
print("p'(1):\t", evaluar(derivar(p), 1))
print("\nPrueba 3")
print("(p + q)(3):\t", evaluar(sumar(p,q), 3))
print("p(3) + q(3):\t", evaluar(p, 3) + evaluar(q, 3))
print("Verificación:\t", math.isclose(evaluar(sumar(p,q), 3), evaluar(p, 3) + evaluar(q, 3)))
````

### Ejercicio 5: Aproximaciones del número $e$

El número $e$ puede aproximarse de dos maneras:

$$S_n = \sum_{k=0}^{n} \frac{1}{k!} \qquad\qquad a_n = \left(1 + \frac{1}{n}\right)^n$$

- Definir `sumas_parciales(N)` que retorne la **lista** `[S_0, S_1, ..., S_{N-1}]`.
- Definir `primer_indice(sumas, tol)` que retorne la **tupla** `(n, S_n, error)` del primer índice cuyo error absoluto respecto a `math.e` sea menor que `tol`, o `None` si ninguno lo cumple.
- Con una **lista por comprensión** crear `tabla`, una lista de tuplas `(n, S_n, error)` para `N = 15`, e imprimir las primeras 8 filas con formato alineado.
- Con otra comprensión, crear `limite`, lista de tuplas `(n, a_n, error)` para `n` en `10, 100, 1000, ..., 10^6`, e imprimirla.
- Imprimir `primer_indice(sumas, 1e-6)` y comparar: ¿cuál de las dos aproximaciones converge más rápido?

````python
# Solución:
def sumas_parciales(N):
    sumas = []
    suma = 0
    for i in range(N):
        suma += 1 / math.factorial(i)
        sumas.append(suma)
    return sumas

def primer_indice(sumas, tol):
    for i in range(len(sumas)):
        error = abs(math.e - sumas[i])
        if error < tol:
            return (i,sumas[i], error)
    return None

sumas = sumas_parciales(15)
tabla = [(i,sumas[i], abs(math.e - sumas[i])) for i in range(15)]

print("Sumas parciales (head 8)")
print("n\tS_n\tError")
for i in range(8):
    print(f"{tabla[i][0]}\t {tabla[i][1]:.3f}\t {tabla[i][2]:.5f}")

limite = [(i, (1+ 1/(10**i))**(10**i), abs(math.e - (1+ 1/(10**i))**(10**i))) for i in range(1,7)]

print("\nLimite")
for i in range(6):
    print(f"10^{limite[i][0]}\t {limite[i][1]:.3f}\t {limite[i][2]:.5f}")

print("\nPrimer Índice sumas parciales")
print(primer_indice(sumas, 1e-6))

# La aproximación por sumas parciales converge más rápido
````
