1. Usando la cadena de caracteres `identidad` definida a continuación, muestre en pantalla el número de caracteres de la cadena y la cadena con la primera letra de cada palabra en mayúscula.

````python
identidad = "la identidad de euler une las constantes e, pi, i, 1 y 0 en una sola ecuación"
````

````python
# Solución
print("Numero de caracteres:", len(identidad))

copia = identidad
res = ""
aux = 0
while aux != 1:
    if copia[:copia.find(" ")][0].isnumeric() != True:  # Identifica solo letras
        res += copia[:copia.find(" ")][0].upper()   # Agrega letra en mayuscula
    if copia.find(" ") == -1:
        aux += 1
    copia = copia[copia.find(" ")+1:]   # Recorta la cadena
print("Cadena con la primera letra de cada palabra en mayuscula:",res)
````

$$
\neg(p \wedge q ) \Longleftrightarrow (\neg p) \vee (\neg q) 
$$
Muestre `ley_simbolos` en pantalla.

*Sugerencia: Aplique `replace` varias veces y tenga cuidado con los espacios alrededor de cada palabra. Para reemplazar usar los codigos únicos de cada símbolo:*

8. La cadena `z` representa un número complejo en la forma `a+bi`. Usando el método `find` y el operador *slicing* (sin usar el método `replace`), construya la cadena `conjugado` que represente al conjugado $\bar{z} = a - bi$ y muéstrela en pantalla.

*Sugerencia: Encuentre la posición del signo `+` y concatene las dos subcadenas que lo rodean junto con el signo `-`.*

````python
z = "12+5i"
````

````python
# Solución
print("COmplejo:", z)
conjugado = z[:z.find("+")] + "-" + z[z.find("+")+1:]
print("Conjugado:", conjugado)
````
